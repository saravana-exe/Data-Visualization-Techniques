```python
import pandas as pd
import sqlite3

df = pd.read_csv("231401501.csv")

conn = sqlite3.connect("231401501.db")

df.to_sql("movies", conn, if_exists="replace", index=False)

query = "SELECT * FROM movies"

df = pd.read_sql(query, conn)

print(df)

conn.close()
```
