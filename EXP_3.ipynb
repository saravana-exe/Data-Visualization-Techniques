```python
import pandas as pd

df = pd.read_csv("231401501.csv")

print(df.isnull())
print(df.isnull().sum())
```

```python
df["release_year"] = df["release_year"].fillna(df["release_year"].mean())
df["worldwide_box_office_millions"] = df["worldwide_box_office_millions"].fillna(
    df["worldwide_box_office_millions"].mean()
)

print(df)
```

```python
print(df.duplicated())

df.drop_duplicates(inplace=True)

print(df)
```

```python
Q1 = df["worldwide_box_office_millions"].quantile(0.25)
Q3 = df["worldwide_box_office_millions"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["worldwide_box_office_millions"] < lower) |
    (df["worldwide_box_office_millions"] > upper)
]

print(outliers)
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["phase"] = encoder.fit_transform(df["phase"])

print(df.head())
```
