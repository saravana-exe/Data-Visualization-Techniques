```python
import pandas as pd

df = pd.read_csv("231401501.csv")

print(df.head())
```

```python
print(df.shape)
```

```python
print(df.dtypes)
```

```python
high_box_office = df[df["worldwide_box_office_millions"] > 1000]

print(high_box_office)
```

```python
phase_3 = df[df["phase"] == "Phase 3"]

print(phase_3)
```

```python
sorted_df = df.sort_values(
    by="worldwide_box_office_millions",
    ascending=False
)

print(sorted_df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

sorted_df = df.sort_values(
    by="worldwide_box_office_millions",
    ascending=False
)

plt.figure(figsize=(12,5))

plt.bar(
    range(len(sorted_df)),
    sorted_df["worldwide_box_office_millions"]
)

plt.title("Worldwide Box Office Ranking")
plt.xlabel("Movies (Sorted)")
plt.ylabel("Worldwide Box Office (in millions)")

plt.show()
```

```python
print("Mean =", df["worldwide_box_office_millions"].mean())
print("Median =", df["worldwide_box_office_millions"].median())
print("Mode =", df["worldwide_box_office_millions"].mode())
print("Standard Deviation =", df["worldwide_box_office_millions"].std())
```
