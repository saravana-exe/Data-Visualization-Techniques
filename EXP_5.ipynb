```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("231401501.csv")

data = df.head()

plt.plot(
    data["release_year"],
    data["worldwide_box_office_millions"]
)

plt.title("Release Year vs Worldwide Box Office")
plt.xlabel("Release Year")
plt.ylabel("Worldwide Box Office")

plt.show()
```

```python
plt.bar(
    data["release_year"],
    data["worldwide_box_office_millions"]
)

plt.title("Release Year vs Worldwide Box Office")
plt.xlabel("Release Year")
plt.ylabel("Worldwide Box Office")

plt.show()
```

```python
phase_count = data["phase"].value_counts()

plt.pie(
    phase_count,
    labels=phase_count.index,
    autopct='%1.1f%%'
)

plt.title("MCU Phase Distribution")

plt.show()
```

```python
plt.boxplot(data["worldwide_box_office_millions"])

plt.title("Worldwide Box Office")
plt.ylabel("Worldwide Box Office")

plt.show()
```
