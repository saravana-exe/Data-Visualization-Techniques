```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/airline_passenger_20000.csv')
```

```python
print(df.shape)
```

```python
print(df.head())
```

```python
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python

df.drop_duplicates(inplace=True)

print("Shape after removing duplicates:", df.shape)
```

```python

print(df.dtypes)
```

```python

service_columns = [
    "Inflight_Wifi_Service",
    "Departure_Arrival_Time_Convenient",
    "Ease_of_Online_Booking",
    "Gate_Location",
    "Food_and_Drink",
    "Online_Boarding",
    "Seat_Comfort",
    "Inflight_Entertainment",
    "On_Board_Service",
    "Leg_Room_Service",
    "Baggage_Handling",
    "Checkin_Service",
    "Inflight_Service",
    "Cleanliness"
]

df["Average_Service_Rating"] = df[service_columns].mean(axis=1)

print(df[["Passenger_ID", "Average_Service_Rating"]].head())
```

```python

print(df.head())
```

```python

print(df.info())
```

```python
print(df.describe())
```

```python
# Display passengers with flight distance greater than 1000
filtered_df = df[df["Flight_Distance"] > 1000]

print(filtered_df)
```

```python
correlation = df[[
    "Age",
    "Flight_Distance",
    "Departure_Delay",
    "Arrival_Delay",
    "Average_Service_Rating"
]].corr()

print(correlation)
```

```python
satisfaction_count = df.groupby("Satisfaction")["Passenger_ID"].count()

# Create bar chart
plt.figure(figsize=(8, 5))
plt.bar(satisfaction_count.index, satisfaction_count.values)

plt.title("Passenger Satisfaction Distribution")
plt.xlabel("Satisfaction")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Flight_Distance"], bins=10, edgecolor="black")

plt.title("Distribution of Flight Distance")
plt.xlabel("Flight Distance")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Flight_Distance"], df["Average_Service_Rating"], alpha=0.6)

plt.title("Flight Distance vs Average Service Rating")
plt.xlabel("Flight Distance")
plt.ylabel("Average Service Rating")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(df["Flight_Distance"])

plt.title("Flight Distance Distribution")
plt.ylabel("Flight Distance")

plt.show()
```

```python
avg_rating = df.groupby("Class")["Average_Service_Rating"].mean()

# Create bar chart
plt.figure(figsize=(8, 5))
plt.bar(avg_rating.index, avg_rating.values)

plt.title("Average Service Rating by Class")
plt.xlabel("Class")
plt.ylabel("Average Service Rating")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(df.index, df["Average_Service_Rating"])

plt.title("Average Service Rating Across Passengers")
plt.xlabel("Record Index")
plt.ylabel("Average Service Rating")

plt.show()
```

```python

df.to_csv("airline_passenger_20000_cleaned.csv", index=False)

print("Cleaned airline passenger dataset exported successfully!")
```
